```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from webdriver_manager.chrome import ChromeDriverManager
import pandas as pd
import time
import os
```

```python
# 드라이버 설정
options = Options()
options.add_argument('--headless')  # 창 없이 실행
options.add_argument('--no-sandbox')
options.add_argument('--disable-dev-shm-usage')
options.add_argument("window-size=1920,1080")
options.add_argument("user-agent=Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 Chrome/121.0.0.0 Safari/537.36")

service = Service(executable_path=ChromeDriverManager().install())
driver = webdriver.Chrome(service=service, options=options)

# Jobkorea 검색 페이지 이동
url = 'https://www.jobkorea.co.kr/Search/?stext=데이터분석'
driver.get(url)
time.sleep(5)

titles = []
companies = []
links = []
details = []


jobs = driver.find_elements(By.CSS_SELECTOR, 'article.list-item')

for job in jobs:
    try:
        a_tag = job.find_element(By.CSS_SELECTOR, 'a.information-title-link')
        title = a_tag.text.strip()
        href = a_tag.get_attribute('href')
        full_link = 'https://www.jobkorea.co.kr' + href if href.startswith('/') else href

        # 회사명
        company = job.find_element(By.CSS_SELECTOR, 'div.list-section-corp').text.strip()

        # 상세조건
        detail_tags = job.find_elements(By.CSS_SELECTOR, 'ul.chip-information-group li.chip-information-item')
        detail_text = ', '.join([d.text.strip() for d in detail_tags if d.text.strip()])

        # 리스트에 추가
        titles.append(title)
        companies.append(company)
        links.append(full_link)
        details.append(detail_text)

    except Exception as e:
        print("오류 발생:", e)
        continue

driver.quit()

# 결과 저장
df = pd.DataFrame({
    'Site': ['Jobkorea'] * len(titles),
    'Col_Company': companies,
    'Col_Recruit': titles,
    'Col_detail': details,
    'Col_url': links
})


df.to_csv('data_tmp/data_jobkorea.csv', index=False, encoding='utf-8-sig')
print("Jobkorea 채용공고 저장 완료!")
display(df.head())
```
