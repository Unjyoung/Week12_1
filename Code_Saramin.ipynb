```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager
import pandas as pd
import time
```

```python
service = Service(executable_path=ChromeDriverManager().install())
driver = webdriver.Chrome(service=service)

url = 'https://www.saramin.co.kr/zf_user/search?search_area=main&searchword=데이터분석'
driver.get(url)

time.sleep(20)

titles = []
companies = []
links = []
details = []

jobs = driver.find_elements(By.CSS_SELECTOR, 'div.area_job')

for job in jobs:
    try:
        a_tag = job.find_element(By.CSS_SELECTOR, 'h2.job_tit a.data_layer')
        title = a_tag.text.strip()
        link = a_tag.get_attribute('href')

        company_elem = job.find_elements(By.CSS_SELECTOR, 'div.area_corp strong.corp_name a')
        company = company_elem[0].text.strip() if company_elem else "정보 없음"

        detail_spans = job.find_elements(By.CSS_SELECTOR, 'div.job_condition span')
        detail_list = [span.text.strip() for span in detail_spans if span.text.strip()]

        titles.append(title)
        links.append(link)
        companies.append(company)
        details.append(detail_list)

    except Exception as e:
        print("오류 발생:", e)
        continue

driver.quit()

df = pd.DataFrame({
    'Site': ['Saramin'] * len(titles),
    'Col_Company': companies,
    'Col_Recruit': titles,
    'Col_detail': details,
    'Col_url': links
})

df.to_csv('data_tmp/data_saramin.csv', index=False, encoding='utf-8-sig')
print(f"Saramin 채용공고 저장 완료!")
display(df.head())
```
